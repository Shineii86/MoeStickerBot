```python
#@title 🎨 0. Load Termux UI Helpers (Run First!)

# 🎨 Terminal Color Palette - Termux/Terminal Style
class TermuxUI:
    RESET = '\033[0m'
    BOLD = '\033[1m'
    DIM = '\033[2m'
    UNDERLINE = '\033[4m'
    
    # Foreground Colors
    BLACK, RED, GREEN, YELLOW = '\033[30m', '\033[31m', '\033[32m', '\033[33m'
    BLUE, MAGENTA, CYAN, WHITE = '\033[34m', '\033[35m', '\033[36m', '\033[37m'
    BRIGHT_BLACK = '\033[90m'
    BRIGHT_RED, BRIGHT_GREEN, BRIGHT_YELLOW = '\033[91m', '\033[92m', '\033[93m'
    BRIGHT_BLUE, BRIGHT_MAGENTA, BRIGHT_CYAN, BRIGHT_WHITE = '\033[94m', '\033[95m', '\033[96m', '\033[97m'
    
    # Background Colors
    BG_BLACK, BG_RED, BG_GREEN, BG_YELLOW = '\033[40m', '\033[41m', '\033[42m', '\033[43m'
    BG_BLUE, BG_MAGENTA, BG_CYAN, BG_WHITE = '\033[44m', '\033[45m', '\033[46m', '\033[47m'
    BG_BRIGHT_BLACK = '\033[100m'
    BG_BRIGHT_BLUE = '\033[104m'
    
    # Prompt Style: ➜ root@shinei:~$
    @staticmethod
    def prompt(user="root", host="shinei", path="~", symbol="➜"):
        return (f"{TermuxUI.BRIGHT_CYAN}{symbol}{TermuxUI.RESET} "
                f"{TermuxUI.BRIGHT_GREEN}{user}@{host}{TermuxUI.RESET}:"
                f"{TermuxUI.BRIGHT_BLUE}{path}{TermuxUI.RESET}$ ")

# Quick access variables for inline usage
G, R, Y, B, M, C, W = (TermuxUI.BRIGHT_GREEN, TermuxUI.BRIGHT_RED, 
                        TermuxUI.BRIGHT_YELLOW, TermuxUI.BRIGHT_BLUE,
                        TermuxUI.BRIGHT_MAGENTA, TermuxUI.BRIGHT_CYAN, 
                        TermuxUI.RESET)
BG_DARK = TermuxUI.BG_BRIGHT_BLACK
BG_INFO = TermuxUI.BG_BRIGHT_BLUE

# Terminal-style print with automatic prompt prefix
def tprint(*args, color=C, prefix=True, end='\n', **kwargs):
    """Terminal-style print with prompt prefix and color support"""
    prompt = TermuxUI.prompt() if prefix else ""
    content = " ".join(str(a) for a in args)
    print(f"{prompt}{color}{content}{W}", end=end, **kwargs)

# Header printer with gradient-style borders
def print_header(title, subtitle="", width=58):
    """Print a Termux-style header with decorative borders"""
    border = f"{B}{'═' * width}{W}"
    print(f"\n{border}")
    print(f"{B}  {C}✨ {title} {W}{B}{' ' * max(0, width - len(title) - 6)}{W}")
    if subtitle:
        print(f"{B}  {DIM}{subtitle}{W}{B}{' ' * max(0, width - len(subtitle) - 4)}{W}")
    print(f"{border}\n")

# Status box printer
def print_status_box(status, message, status_color=G):
    """Print a status indicator with colored badge"""
    print(f"{B}[{status_color}{status}{W}{B}]{W} {message}")

# Animated spinner for waiting operations
def spinner_task(duration, message, success_msg="done"):
    """Terminal-style animated spinner"""
    import sys, time
    spinners = ['⠋', '⠙', '⠹', '⠸', '⠼', '⠴', '⠦', '⠧', '⠇', '⠏']
    end_time = time.time() + duration
    i = 0
    print(f"{TermuxUI.prompt()}{C}⏳ {message}{W} ", end="", flush=True)
    while time.time() < end_time:
        sys.stdout.write(f"\r{TermuxUI.prompt()}{C}{spinners[i % len(spinners)]} {message}{W} ")
        sys.stdout.flush()
        i += 1
        time.sleep(0.08)
    sys.stdout.write(f"\r{TermuxUI.prompt()}{G}✓{W} {message} {G}{success_msg}{W}\n")
    sys.stdout.flush()

# Clear terminal display
def clear_terminal():
    print("\033[2J\033[H", end="")

# Initialize UI
print(f"{BG_DARK}{BRIGHT_WHITE} 🎨 Termux UI Loaded {W}{BG_DARK}{' ' * 25}{W}")
tprint("ANSI colors: enabled", color=G)
tprint("Prompt style: ➜ root@shinei:~$", color=C)
tprint("Background highlights: enabled", color=M)
print(f"{B}{'─' * 58}{W}\n")
```

```python
#@title 📦 2. Install System Dependencies [TERMUX UI]

import subprocess, sys, os, urllib.request, time
from tqdm.notebook import tqdm

print_header("📦 SYSTEM DEPENDENCIES", "Installing core tools for sticker processing")

steps = [
    ("🔄 Updating package lists", "apt-get update -qq"),
    ("📦 Installing imagemagick, ffmpeg, exiv2...", 
     "apt-get install -y -qq imagemagick libarchive-tools ffmpeg curl gifsicle python3 exiv2"),
]

for desc, cmd in tqdm(steps, desc=f"{C}System Setup{W}", unit="step", 
                       bar_format=f"{B}{{l_bar}}{C}{{bar}}{B}{{r_bar}}{W}"):
    tprint(f"⏳ {desc}", color=C)
    result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    if result.returncode == 0:
        print(f"           {G}✓{W} Completed\n")
    else:
        print(f"           {R}✗{W} Failed - check logs\n")

# Go installation with progress
tprint("⏳ Downloading Go toolchain...", color=C)
url = "https://go.dev/dl/go1.21.5.linux-amd64.tar.gz"
filename = "go1.21.5.linux-amd64.tar.gz"

with tqdm(unit='B', unit_scale=True, unit_divisor=1024, miniters=1, 
          desc=f"{B}📥{W} Go", bar_format=f"{B}{{l_bar}}{C}{{bar}}{B}{{r_bar}}{W}") as t:
    urllib.request.urlretrieve(url, filename, 
        reporthook=lambda b, bsize, tsize: t.update(b*bsize - t.n))

spinner_task(2, "Extracting Go")
!tar -C /usr/local -xzf go1.21.5.linux-amd64.tar.gz > /dev/null 2>&1

# Setup environment
os.environ['PATH'] += ":/usr/local/go/bin"
os.environ['GOPATH'] = "/root/go"
os.environ['GO111MODULE'] = "on"
!mkdir -p $GOPATH

# Final status report
print(f"\n{BG_DARK}{B} 🎉 INSTALLATION COMPLETE {W}{BG_DARK}{' ' * 20}{W}")
tprint("✓ All dependencies ready!", color=G, prefix=False)
print(f"{B}{'─' * 58}{W}")
print(f"{B}  {C}INSTALLED VERSIONS{W}")
print(f"{B}{'─' * 58}{W}")

tools = [
    ("Go", "go version"),
    ("exiv2", "exiv2 --version | head -1"),
    ("ffmpeg", "ffmpeg -version | head -1"),
    ("ImageMagick", "convert --version | head -1"),
]
for name, cmd in tools:
    try:
        if "exiv2" in cmd:
            out = subprocess.getoutput(cmd).split()[-1]
        else:
            out = subprocess.getoutput(cmd).split()[2]
        print(f"  {G}• {name:12}{W} {C}{out}{W}")
    except:
        print(f"  {Y}• {name:12}{W} {DIM}checking...{W}")
print(f"{B}{'─' * 58}{W}\n")
```

```python
#@title 🐍 3. Install Python Helper Scripts [TERMUX UI]

print_header("🐍 PYTHON HELPERS", "Downloading utility scripts")

helpers = [
    ("msb_emoji.py", "Emoji processing utilities"),
    ("msb_kakao_decrypt.py", "KakaoTalk pack decryption"),
    ("msb_rlottie.py", "Lottie animation support")
]

for filename, desc in tqdm(helpers, desc=f"{C}Downloading{W}", unit="file",
                           bar_format=f"{B}{{l_bar}}{C}{{bar}}{B}{{r_bar}}{W}"):
    tprint(f"⏳ {desc}", color=C)
    !wget -q https://raw.githubusercontent.com/Shineii86/MoeStickersBot/master/tools/{filename} -O /usr/local/bin/{filename}
    !chmod +x /usr/local/bin/{filename}
    print(f"           {G}✓{W} {filename} installed\n")

print(f"\n{BG_DARK}{B} ✅ HELPERS READY {W}{BG_DARK}{' ' * 35}{W}")
tprint(f"All scripts installed to {C}/usr/local/bin/{W}", prefix=False)
```

```python
#@title 🔨 4. Build MoeStickersBot from Source [TERMUX UI]

print_header("🔨 BUILDING BOT", "Compiling from GitHub source")

!rm -rf MoeStickersBot

tprint("⏳ Cloning repository...", color=C)
with tqdm(total=100, desc=f"{B}📥{W} Clone", unit="%", 
          bar_format=f"{B}{{l_bar}}{C}{{bar}}{B}{{r_bar}}{W}") as pbar:
    !git clone --depth 1 --quiet https://github.com/Shineii86/MoeStickersBot.git
    pbar.update(100)
print(f"           {G}✓{W} Repository cloned\n")

%cd MoeStickersBot

tprint("⏳ Downloading Go modules...", color=C)
!go mod download > /dev/null 2>&1
print(f"           {G}✓{W} Modules cached\n")

tprint("⏳ Compiling bot binary...", color=C)
spinner_task(45, "Building", success_msg="compiled")
!go build -o MoeStickersBot cmd/MoeStickersBot/main.go 2>&1 | grep -v "^#" || true

if os.path.exists("MoeStickersBot"):
    size = os.path.getsize("MoeStickersBot") / (1024 * 1024)
    print(f"\n{BG_DARK}{B} 🎉 BUILD SUCCESS {W}{BG_DARK}{' ' * 38}{W}")
    tprint(f"✓ Binary ready: {G}{size:.1f} MB{W}", prefix=False)
else:
    print(f"\n{R}❌ Build failed. Check output above.{W}")
```

```python
#@title ⚙️ 5. Bot Configuration [TERMUX UI]

print_header("⚙️ CONFIGURATION", "Set up your bot credentials")

# ========== REQUIRED ==========
BOT_TOKEN = ""  #@param {type:"string"}

# ========== DATABASE (OPTIONAL) ==========
ENABLE_DB = False  #@param {type:"boolean"}
DB_ADDR = "localhost:3306"  #@param {type:"string"}
DB_USER = "moe_bot"  #@param {type:"string"}
DB_PASS = ""  #@param {type:"string"}
DB_NAME = "moe_sticker_bot"  #@param {type:"string"}

# ========== WEBAPP (OPTIONAL - Requires ngrok) ==========
ENABLE_WEBAPP = False  #@param {type:"boolean"}
WEBAPP_PORT = 8080  #@param {type:"integer"}
NGROK_AUTHTOKEN = ""  #@param {type:"string"}
# Get your free ngrok auth token at https://dashboard.ngrok.com/auth

# ========== GENERAL ==========
DATA_DIR = "moe_sticker_bot_data"  #@param {type:"string"}
LOG_LEVEL = "info"  #@param ["debug", "info", "warn", "error"]
HTTP_PROXY = ""  #@param {type:"string"}

print(f"{B}{'─' * 58}{W}")
if BOT_TOKEN:
    masked = BOT_TOKEN[:8] + "..." + BOT_TOKEN[-4:]
    print_status_box("✓", f"Bot Token: {C}{masked}{W}", G)
else:
    print_status_box("⚠", f"Please enter your Bot Token above", Y)

if ENABLE_DB:
    print_status_box("✓", f"Database: {C}{DB_ADDR}/{DB_NAME}{W}", B)
else:
    print_status_box("○", f"Database: {DIM}disabled{W}", TermuxUI.BRIGHT_BLACK)

if ENABLE_WEBAPP:
    if NGROK_AUTHTOKEN:
        print_status_box("✓", f"WebApp: {C}enabled (ngrok){W}", M)
    else:
        print_status_box("⚠", f"WebApp enabled but no ngrok token", Y)
        ENABLE_WEBAPP = False
else:
    print_status_box("○", f"WebApp: {DIM}disabled{W}", TermuxUI.BRIGHT_BLACK)

print(f"{B}{'─' * 58}{W}\n")

if ENABLE_WEBAPP and not NGROK_AUTHTOKEN:
    tprint("⚠️  WebApp is enabled but no ngrok auth token provided. WebApp will be disabled.", color=Y, prefix=False)
```

```python
#@title 🌐 5.5 Setup ngrok Tunnel (if WebApp enabled) [TERMUX UI]

import subprocess, json, requests, time, sys

WEBAPP_URL = ""
ngrok_process = None

if ENABLE_WEBAPP:
    print_header("🌐 NGROK TUNNEL", "Exposing WebApp to the internet")

    # Download ngrok if not present
    if not os.path.exists("./ngrok"):
        tprint("⏳ Downloading ngrok...", color=C)
        !wget -q https://bin.equinox.io/c/bNyj1mQVY4c/ngrok-v3-stable-linux-amd64.tgz
        !tar -xzf ngrok-v3-stable-linux-amd64.tgz > /dev/null 2>&1
        !chmod +x ngrok
        print(f"           {G}✓{W} ngrok downloaded\n")

    # Authenticate
    tprint("⏳ Authenticating with ngrok...", color=C)
    !./ngrok config add-authtoken {NGROK_AUTHTOKEN} > /dev/null 2>&1
    print(f"           {G}✓{W} Authenticated\n")

    # Kill any previous ngrok instances
    !pkill -f ngrok || true

    # Start ngrok in background
    tprint(f"⏳ Starting tunnel on port {WEBAPP_PORT}...", color=C)
    ngrok_process = subprocess.Popen(
        ["./ngrok", "http", str(WEBAPP_PORT), "--log", "stdout"],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    )

    # Wait for tunnel to be ready and get public URL
    spinner_task(5, "Establishing tunnel")
    max_retries = 10
    for i in range(max_retries):
        try:
            resp = requests.get("http://127.0.0.1:4040/api/tunnels", timeout=2)
            if resp.status_code == 200:
                tunnels = resp.json()['tunnels']
                if tunnels:
                    WEBAPP_URL = tunnels[0]['public_url']
                    print(f"\n{BG_DARK}{B} 🔗 TUNNEL ACTIVE {W}{BG_DARK}{' ' * 37}{W}")
                    tprint(f"✓ Public URL: {C}{WEBAPP_URL}{W}", prefix=False)
                    break
        except:
            pass
        time.sleep(1)
    else:
        print(f"\n{R}❌ Failed to retrieve ngrok URL. WebApp will be disabled.{W}")
        ENABLE_WEBAPP = False
else:
    tprint("ℹ️  WebApp not enabled. Skipping ngrok setup.", color=TermuxUI.BRIGHT_BLACK)
```

```python
#@title 🚀 6. Launch Bot (Background) [TERMUX UI]

import subprocess, sys, time

print_header("🚀 LAUNCHING BOT", "Starting MoeStickersBot daemon")

if not BOT_TOKEN:
    print(f"{R}❌{W} No Bot Token provided. Please enter it in the Configuration cell.")
    sys.exit(1)

cmd = [
    "./MoeStickersBot",
    f"--bot_token={BOT_TOKEN}",
    f"--log_level={LOG_LEVEL}",
    f"--data_dir={DATA_DIR}"
]

if ENABLE_DB and DB_ADDR:
    cmd.extend([
        f"--db_addr={DB_ADDR}",
        f"--db_user={DB_USER}",
        f"--db_pass={DB_PASS}",
        f"--db_name={DB_NAME}"
    ])

if ENABLE_WEBAPP and WEBAPP_URL:
    cmd.append(f"--webapp_url={WEBAPP_URL}")
    cmd.append(f"--webapp_listen_addr=0.0.0.0:{WEBAPP_PORT}")

if HTTP_PROXY:
    cmd.append(f"--http_proxy={HTTP_PROXY}")

cmd_display = ' '.join(cmd).replace(BOT_TOKEN, '[SECRET]')
tprint(f"🔧 Command: {C}{cmd_display}{W}", prefix=False)
print()

log_stdout = open("bot_stdout.log", "w")
log_stderr = open("bot_stderr.log", "w")
process = subprocess.Popen(cmd, stdout=log_stdout, stderr=log_stderr)

time.sleep(5)
if process.poll() is None:
    print(f"\n{BG_DARK}{B} 🎉 BOT RUNNING {W}{BG_DARK}{' ' * 40}{W}")
    tprint(f"✅ Bot is {G}RUNNING{W}!", prefix=False)
    tprint(f"📱 Go to Telegram and send {C}/start{W} to your bot!", prefix=False)
    if ENABLE_WEBAPP and WEBAPP_URL:
        tprint(f"🌐 WebApp available at: {C}{WEBAPP_URL}{W}", prefix=False)
    tprint(f"📋 Logs: {C}bot_stdout.log{W}, {C}bot_stderr.log{W}", prefix=False)
    print(f"\n{B}{'─' * 58}{W}")
    print(f"{B}  {DIM}Tip: Use the 'View Live Logs' cell to monitor output{W}")
    print(f"{B}{'─' * 58}{W}\n")
else:
    print(f"\n{R}❌ Bot exited unexpectedly. Check stderr log:{W}")
    !cat bot_stderr.log
```

```python
#@title 📜 View Live Logs [TERMUX UI]
LOG_TYPE = "stderr"  #@param ["stdout", "stderr"]
LINES = 30  #@param {type:"slider", min:10, max:100, step:10}
log_file = f"bot_{LOG_TYPE}.log"

print_header(f"📜 LOG VIEWER", f"Showing last {LINES} lines of {log_file}")

# Color-code log levels in output
import re
def colorize_log(line):
    if "error" in line.lower() or "err" in line.lower():
        return f"{R}{line}{W}"
    elif "warn" in line.lower():
        return f"{Y}{line}{W}"
    elif "debug" in line.lower():
        return f"{DIM}{line}{W}"
    return line

print(f"{B}{'─' * 58}{W}")
!tail -n {LINES} {log_file} 2>/dev/null | while read line; do echo -e "$line"; done
print(f"{B}{'─' * 58}{W}\n")

tprint(f"💡 Tip: Switch LOG_TYPE to view stdout/stderr", color=TermuxUI.BRIGHT_BLACK)
```

```python
#@title 🛑 Stop Bot & ngrok [TERMUX UI]

print_header("🛑 SHUTDOWN", "Stopping services cleanly")

tprint("⏳ Stopping bot process...", color=C)
result = !pkill -f MoeStickersBot && echo "stopped" || echo "not_running"
if "stopped" in result[0]:
    print(f"           {G}✓{W} Bot stopped\n")
else:
    print(f"           {Y}○{W} No bot running\n")

tprint("⏳ Stopping ngrok tunnel...", color=C)
result = !pkill -f ngrok && echo "stopped" || echo "not_running"
if "stopped" in result[0]:
    print(f"           {G}✓{W} ngrok stopped\n")
else:
    print(f"           {Y}○{W} No ngrok running\n")

print(f"{BG_DARK}{B} ✅ CLEAN SHUTDOWN {W}{BG_DARK}{' ' * 35}{W}")
tprint("All services terminated safely", color=G, prefix=False)
```
